```python
import pandas as pd
from sqlalchemy import create_engine
from getpass import getpass
from urllib.parse import quote_plus

pwd = getpass("PostgreSQL password: ")
engine = create_engine(
    f"postgresql+psycopg2://postgres:{quote_plus(pwd)}@localhost:5432/covid_dw"
)
```

```python
query = """
SELECT t.date, p.pays, p.population,
       f.nouveaux_cas, f.nouveaux_deces,
       f.cas_cumules, f.deces_cumules,
       f.cas_hebdomadaires, f.deces_hebdomadaires
FROM fait_pandemie f
JOIN dim_temps t ON t.id_temps = f.id_temps
JOIN dim_pays  p ON p.id_pays  = f.id_pays
"""
df = pd.read_sql(query, engine)
df["date"] = pd.to_datetime(df["date"])
print(df.shape)
df.head()
```

```python
print(df.loc[df["date"] == "2023-03-09", "cas_cumules"].sum())
print(df["pays"].nunique())
```
